```python
from itkwidgets import view
import nibabel as nib
import numpy as np
from scipy.ndimage import zoom

def view_3d_itk(image_path, label_path, modality_index=0, downscale_factor=0.5):
    img = nib.load(image_path).get_fdata()
    seg = nib.load(label_path).get_fdata()

    img = img[..., modality_index]
    img = (img - np.min(img)) / (np.max(img) - np.min(img) + 1e-8)
    img = img.astype(np.float32)
    seg = seg.astype(np.uint8)

    if downscale_factor < 1.0:
        zoom_factors = [downscale_factor] * 3
        img = zoom(img, zoom_factors, order=1)
        seg = zoom(seg, zoom_factors, order=0)  # nearest for labels

    print("✅ Shapes after scaling:", img.shape, seg.shape)
    view(image=img, label_image=seg)
```

```python
view_3d_itk(
    image_path='./dataset/imagesTr/BRATS_001.nii.gz',
    label_path='./dataset/labelsTr/BRATS_001.nii.gz',
    modality_index=1,         # e.g., FLAIR
    downscale_factor=0.5      # or 0.3 for more safety
)
```
